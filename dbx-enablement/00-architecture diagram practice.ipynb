## The Diagram, Column by Column

| **Sources** | Sensors, ERP (SAP), MES, maintenance logs | Where the data lives today. How each one sends data shapes the design. |

## Tradeoffs to Mention
- **Streaming vs. batch:** Streaming is fast but costs more, so use it only for sensors.
- **CDC vs. full copy:** CDC is lighter on SAP; a full copy is simpler but slower.
- **3 layers vs. fewer:** Each layer has one job (keep, clean, serve). For a simple use case, you could collapse them.
- **Alerts:** Start as recommendations, not automatic shutdowns. Track false alarms and retrain.

## Batch vs. Streaming

**Batch:** process data in groups, on a schedule
- Runs hourly, nightly, etc.
- Data is minutes to hours old
- Cheaper and simpler
- Good for: daily reports, dashboards
- In my diagram: ERP, MES, maintenance logs

**Streaming:** process data continuously, as it arrives
- Runs all the time
- Data is seconds old
- More expensive and complex
- Good for: fraud alerts, machine failure alerts
- In my diagram: IoT sensors

**How to choose:** How fast does someone need to act on the data?
- Right away → **Streaming**
- Tomorrow is fine → **Batch**

> **Interview line:** "I use streaming only where speed changes the outcome, like catching a failing machine. Everything else runs as batch because it's cheaper and simpler."
