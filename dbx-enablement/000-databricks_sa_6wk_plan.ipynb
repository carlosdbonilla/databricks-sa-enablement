# Databricks Senior SA: 6-Week Learning Plan

**Goal:** be able to whiteboard an end-to-end data and AI architecture for a regulated-industry customer, explain every component in plain English, and tie it to business value.

**Pace:** about 15-18 hours a week. Courses are the smallest share of your time. Architects are made by designing, building and explaining, not by watching videos.

| Activity | Share of time | What it builds |
| --- | --- | --- |
| Design reps | 30% | Thinking like an architect under interview pressure |
| Hands-on building | 30% | Real confidence, from fixing real errors |
| Understanding the why | 20% | Trade-offs: why a customer picks one option, and the downside |
| Courses and certifications | 20% | Structure, breadth and a credibility signal |

**Weekly schedule:**
- Monday, Tuesday, Thursday, Friday: 1.5 hours of course content
- Wednesday: 1.5-hour design rep (from Week 2)
- Saturday: 4-hour hands-on build in Databricks Free Edition. Never skip it.
- Sunday: 2-hour design rep, then a 30-minute review of the week

**Every topic goes through four steps:**
1. **Learn:** watch the module or read the chapter
2. **Build:** do it yourself in a real workspace
3. **Explain:** write a short note here, then explain it out loud in 2 minutes as if to a customer
4. **Test:** take the quiz or a practice exam

**Rules:** never skip two days in a row; block the hours on your calendar; if you can't explain a term in one sentence, add it to the glossary.

**Habits of a seasoned architect:**
- Explain everything out loud. If you can't explain it simply to a non-technical executive, you don't understand it yet.
- Ask "why" for every feature: why would a customer choose this, and what's the downside?
- Keep a pattern library from Week 2: every architecture you meet (fraud detection, customer 360, claims, warehouse migration) with its industry, data sources, components and business metric.
- Talk to practitioners: Databricks community forums, events like the Data + AI World Tour, and SAs on LinkedIn.

## Week 1: Cloud, data and API foundations

- **Cloud:** AWS Cloud Practitioner Essentials (digital course on AWS Skill Builder). Do only: cloud basics, global infrastructure, compute (EC2), networking (VPC), storage and databases (S3), security (IAM), pricing and billing including egress. Skip support plans, migration frameworks and the AI/ML module. https://skillbuilder.aws/learn/94T2BEN85A/aws-cloud-practitioner-essentials/8D79F3AVR7 
- **Data:** [Get Started With Data Engineering on Databricks](https://www.databricks.com/resources/ebook/get-started-data-engineering-databricks) (free, about 2 hours, with a certificate)
- **Weekend reading:** skim the intro and industry chapters of the [Big Book of Data Engineering, 4th Edition](https://entrepreneur.tradepub.com/free/w_daac113/) (free Databricks eBook)
- **Gaps to fill with the glossary:** warehouse vs lake vs lakehouse, OLTP vs OLAP, batch vs streaming, ETL vs ELT, Parquet, Kafka, CDC
- **APIs:** call a free public API with Postman or Python `requests`
- **SQL and Python:** SELECT, JOIN, GROUP BY; load a CSV with pandas
- **Accreditation:** Databricks Fundamentals (free, about 1.5 hours), if you haven't earned it yet
- **Build:** AWS free-tier account, S3 bucket, upload a CSV

**Gate:** explain every glossary term out loud without notes.

## Week 2: Databricks core platform

Main course: the **Data Engineer Associate learning pathway** in [Databricks Academy](https://www.databricks.com/learn/training/home) (all self-paced courses are free). Start with the free Academy course **Get Started with Databricks for Data Engineering**. Do every lab in [Databricks Free Edition](https://www.databricks.com/learn/free-edition).

- **Lakehouse:** one copy of data serving BI, ML and AI
- **Delta Lake:** ACID transactions, time travel, schema enforcement
- **Unity Catalog:** catalog > schema > table, permissions, lineage, PII masking
- **Medallion architecture:** bronze (raw), silver (cleaned), gold (business-ready)
- **Ingestion:** Auto Loader (files), Lakeflow Connect (SaaS and databases), Lakeflow Declarative Pipelines
- **Build (capstone part 1):** fraud dataset through bronze, silver and gold, with PII masked in Unity Catalog

**Design reps:** scenarios 1 and 2. Start your pattern library.

## Week 3: Operations, analytics and first certification

- **Orchestration:** Lakeflow Jobs; CI/CD with Databricks Asset Bundles
- **Compute and cost:** classic clusters, serverless, SQL warehouses, DBU pricing
- **Analytics:** Databricks SQL, AI/BI dashboards, Genie spaces
- **Build (capstone part 2):** schedule the pipeline as a job; dashboard and Genie space on the gold tables
- **Exam:** two practice exams, then sit the **Data Engineer Associate** ($200, 45 scored questions in 90 minutes)

**Design reps:** scenarios 3 and 4.

**Gate:** Data Engineer Associate passed.

## Week 4: ML and GenAI on Databricks

- **Classical ML:** MLflow tracking, model registry in Unity Catalog, Model Serving. Use cases: fraud, churn, forecasting, credit risk.
- **GenAI course:** **Generative AI Engineering with Databricks** (RAG, agents, evaluation and governance, deployment and monitoring)
- **Concepts:** tokens, embeddings, Vector Search, RAG, agents and tools, AI Functions, Agent Bricks, LLM judges, guardrails, inference tables, Unity AI Gateway
- **Build (capstone part 3):** fraud model in MLflow, plus a RAG assistant over sample compliance policy PDFs

**Design reps:** scenarios 5 and 6. **Accreditation:** Generative AI Fundamentals (free, about 1 hour).

**Gate:** explain RAG vs fine-tuning, and how Unity Catalog governs AI, in 2 minutes each. The GenAI Engineer Associate exam ([exam guide](https://www.databricks.com/sites/default/files/2026-02/Databricks-Certified-Generative-AI-Engineer-Associate-Exam-Guide-Interrim-Feb26.pdf)) can come after week 6.

## Week 5: Industry use cases and positioning

- **Regulated industries:** financial services (fraud, AML, risk), healthcare (patient analytics, claims), public sector. Why they care about governance, audit and data residency.
- **Study:** [Architecture Center](https://www.databricks.com/resources/architectures), [Solution Accelerators](https://www.databricks.com/solutions/accelerators), 10 customer stories in your verticals. Add each pattern to your library.
- **Competitors:** Snowflake and Microsoft Fabric, the two you'll hear about most. One strength of each and the Databricks counter. Migration from legacy warehouses and Hadoop.
- **Design reps:** scenarios 7 and 8
- **Build (capstone part 4):** architecture diagram and business value story
- **Optional, first to drop if time is short:** AWS Platform Architect accreditation (customer-managed VPCs and keys, external storage, cloud integrations)

## Week 6: Design interview drills

**The seven-step design framework:**
1. **Discovery:** business goal, users, data sources, volume, latency, compliance
2. **Ingest:** batch or streaming, and which tool
3. **Store and process:** medallion layers on Delta Lake
4. **Govern:** Unity Catalog permissions, lineage, PII masking
5. **Serve:** BI, ML, GenAI, apps
6. **Operate:** orchestration, CI/CD, monitoring, cost
7. **Value:** tie back to the customer's metric and MEDDPICC

**Reference architecture to whiteboard from memory (under 5 minutes):**

```
SOURCES            INGEST                 STORE AND PROCESS          SERVE
Databases (CDC) -> Lakeflow Connect  \                           /-> BI and Genie
SaaS apps, APIs -> Lakeflow Connect   \   Lakehouse on Delta     |-> ML models
Files in S3     -> Auto Loader      -----> Bronze -> Silver -> Gold -> GenAI and agents
Kafka streams   -> Spark Streaming    /                          \-> Apps and sharing

[ Unity Catalog governs every layer: access, lineage, PII masking, audit ]
[ Lakeflow Jobs orchestrate, Asset Bundles handle CI/CD, monitoring tracks cost ]
```

**Mock scenarios (45 minutes each, recorded):**
- A bank wants real-time card fraud detection
- A hospital wants to predict patient readmissions
- An insurer wants a GenAI assistant for claims adjusters
- A retailer wants demand forecasting across 500 stores

**Capstone part 5:** record a 5-minute demo, post the code on GitHub and the demo on LinkedIn.

**Gate:** deliver a 45-minute mock design to a peer or mentor without notes.

## Glossary

| Term | Plain English | Why customers care |
| --- | --- | --- |
| Region / availability zone | A geographic area / a separate data center inside it | Latency, resilience, regulations |
| Object storage (S3, ADLS, GCS) | Cheap, near-unlimited storage for files | Where lakehouse data sits |
| Compute (EC2, VMs) | Rented servers that run your code | Biggest cloud cost driver |
| IAM | Who can do what in the cloud account | Security and audit |
| VPC | A private network in the cloud | Regulated customers require it |
| Egress cost | Fee charged when data leaves the cloud network or region | Copying data to other tools gets expensive |
| API | A defined way for two programs to talk | How systems connect |
| REST / JSON | Most common API style / its text data format | How most SaaS data arrives |
| Batch vs streaming | Scheduled chunks vs continuous processing | Fraud needs seconds; reports can wait |
| Kafka | Carries a continuous stream of events between apps | Common real-time source |
| CDC | Capturing only the database rows that changed | Fresh data without full reloads |
| ETL vs ELT | Transform before loading vs after loading | ELT fits the lakehouse |
| Data warehouse | Structured, SQL-optimized store for reporting | Often what customers migrate from |
| Data lake | Raw files of any type in cheap storage | Flexible but messy without governance |
| Lakehouse | Data lake with warehouse reliability and governance | One platform for BI, ML, AI |
| Parquet / Delta Lake | Columnar file format / Parquet plus transactions and history | Open formats, no lock-in |
| Apache Spark | Processes huge data across many machines in parallel | Scale |
| Medallion architecture | Bronze, silver, gold layers | Quality and traceability |
| Unity Catalog | Databricks' governance layer for data and AI | Access, lineage, audit in one place |
| Orchestration | Scheduling and chaining jobs in order | Reliable pipelines |
| CI/CD | Automated testing and deployment of code | Fewer production errors |
| DBU | Databricks' billing unit for compute | How the bill is calculated |
| PII | Data that identifies a person | Must be masked by law |
| Embedding / vector search | Text as numbers / search by meaning | Foundation of RAG |
| RAG | Retrieve documents and add them to the AI prompt | Accurate AI on private data |
