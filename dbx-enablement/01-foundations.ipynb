## 🧭 Unity Catalog Hierarchy
```
catalog ➔ schema ➔ table / view / volume / model / function
```
- **Referencing in code:** `catalog.schema.object` (the three-level namespace)
- **Every catalog has 2 standard schemas:**
  - `default`: an empty starter schema
  - `information_schema`: metadata views about everything in the catalog

💬 **Customer framing:** "One governed place to find and secure all your data and AI assets."

## 🗂️ Workspace vs. Catalog Explorer
| Location | What lives there |

## 📓 Notebook Basics
- **Free Edition:** only serverless compute is available, and it starts automatically when you run a cell
- **Mixing languages:** changing one cell's language adds a magic command (`%py`, `%sql`, `%md`). The notebook's default language stays the same.
- **Databricks Assistant:** in a Python notebook, it may answer in PySpark. Say *"in SQL"* in your prompt if you want plain SQL.

## 🔐 Permissions (Hierarchical)
**Gotcha:** `SELECT` on a table is **not enough** to query it. The user also needs:
1. `USE CATALOG` on the parent catalog
2. `USE SCHEMA` on the parent schema

Think of it this way: you need a key to the building, then the floor, then the room.

````sql
%sql
-- Where am I? This tells you whether you need full table names
SELECT current_catalog(), current_schema();

-- Change your context
-- USE CATALOG my_catalog;
-- USE SCHEMA my_schema;

-- Grant access the right way (all three levels)
-- GRANT USE CATALOG ON CATALOG my_catalog TO `user@company.com`;
-- GRANT USE SCHEMA  ON SCHEMA  my_catalog.my_schema TO `user@company.com`;
-- GRANT SELECT      ON TABLE   my_catalog.my_schema.my_table TO `user@company.com`;
````
